```python
my_list = ['apple', 'banana', 'fish', 'mouse']
def make_plural(words):
    result = []

    for i in words:
        if i == 'mouse':
            result.append('mice')
        elif i == 'fish':
            result.append('fishes')
        else:
            result.append(i + 's')

    return result
```
